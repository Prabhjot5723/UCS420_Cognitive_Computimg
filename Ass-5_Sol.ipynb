## Q1. Basic NumPy Operations

```python
import numpy as np

arr = np.array([1, 2, 3, 4, 5])

print("Add 2:", arr + 2)
print("Multiply by 3:", arr * 3)
print("Divide by 2:", arr / 2)
```

## Q2(a). Reverse Array

```python
import numpy as np

arr = np.array([1, 2, 3, 6, 4, 5])
print("Reversed:", arr[::-1])
```

## Q2(b). Most Frequent Value and Indices

```python
import numpy as np

x = np.array([1,2,3,4,5,1,2,1,1,1])
y = np.array([1,1,1,2,3,4,2,4,3,3])

for a in [x, y]:
    values, counts = np.unique(a, return_counts=True)
    most = values[np.argmax(counts)]
    print("Array:", a)
    print("Most frequent value:", most)
    print("Indices:", np.where(a == most)[0])
    print()
```

## Q3. Access Elements

## Q4. Linspace, Properties and Transpose

```python
import numpy as np

Prabhjot = np.linspace(10, 100, 25)

print("Array:", Prabhjot)
print("Dimensions:", Prabhjot.ndim)
print("Shape:", Prabhjot.shape)
print("Total elements:", Prabhjot.size)
print("Data type:", Prabhjot.dtype)
print("Total bytes:", Prabhjot.nbytes)

# Transpose using reshape()
transpose = Prabhjot.reshape(25, 1)
print("Transpose using reshape():")
print(transpose)

# Yes, T can also be used.
print("Transpose using T:")
print(Prabhjot.reshape(1, 25).T)
```

## Q5. Statistics, Reshape and Resize

```python
import numpy as np

ucs420_Prabhjot = np.array([
    [10,20,30,40],
    [50,60,70,80],
    [90,15,20,35]
])

print("Mean:", np.mean(ucs420_Prabhjot))
print("Median:", np.median(ucs420_Prabhjot))
print("Max:", np.max(ucs420_Prabhjot))
print("Min:", np.min(ucs420_Prabhjot))
print("Unique elements:", np.unique(ucs420_Prabhjot))

reshaped_ucs420_Prabhjot = ucs420_Prabhjot.reshape(4,3)
print("Reshaped array:")
print(reshaped_ucs420_Prabhjot)

resized_ucs420_Prabhjot = np.resize(ucs420_Prabhjot, (2,3))
print("Resized array:")
print(resized_ucs420_Prabhjot)
```
