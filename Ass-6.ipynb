## Q1. Sensor Readings

```python
temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

print("Original temperatures:", temperature)

corrected_temperature = temperature + 2
print("\na. After adding 2°C:", corrected_temperature)

fahrenheit = (9/5) * temperature + 32
print("\nb. Temperature in Fahrenheit:", fahrenheit)

greater_than_32 = temperature[temperature > 32]
print("\nc. Readings greater than 32°C:", greater_than_32)

count = np.sum(temperature > 32)
print("\nd. Number of readings greater than 32°C:", count)

print("\ne. Vectorization performs operations on the whole array at once,")
print("so it is faster and simpler than using a for loop.")
print("Boolean indexing directly selects values that satisfy a condition.")
```

## Q2. Daily Steps Matrix

```python
steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

print("Steps matrix:\n", steps)

print("\na. Total steps:", np.sum(steps))

print("b. Mean steps:", np.mean(steps))

print("c. Maximum steps:", np.max(steps))
print("   Minimum steps:", np.min(steps))

print("d. Total steps for each day:", np.sum(steps, axis=0))

print("e. Total steps for each user:", np.sum(steps, axis=1))

max_position = np.unravel_index(np.argmax(steps), steps.shape)
print("f. Position of maximum steps (user, day):", max_position)
print("   Maximum value:", steps[max_position])
```

## Q3. NumPy Arrays, Slicing, Copying and Reshaping

```python
original = np.array([1, 2, 3, 4, 5, 6])
print("a. Original:", original)

subset = original[1:5]
print("\nb. Subset:", subset)

subset[0] = 999
print("\nc. After modifying subset:")
print("Original:", original)
print("Subset:", subset)

copied_array = original[1:5].copy()
copied_array[0] = 500

print("\nd. After modifying copied array:")
print("Original:", original)
print("Copied array:", copied_array)

matrix = np.arange(1, 13).reshape(3, 4)
print("\ne. 3 x 4 Matrix:\n", matrix)

print("\nf. First row:", matrix[0])
print("Last row:", matrix[-1])
print("Second column:", matrix[:, 1])
print("Rows 1-2 and columns 2-3:\n", matrix[1:3, 2:4])

flattened = matrix.flatten()
raveled = matrix.ravel()

print("\ng. Using flatten():", flattened)
print("Using ravel():", raveled)

raveled[0] = 100
print("\nh. After modifying ravel()[0] = 100:")
print("Original matrix:\n", matrix)
print("The original changes because ravel() usually returns a view.")

flattened[1] = 200
print("\ni. After modifying flatten()[1] = 200:")
print("Original matrix:\n", matrix)
print("Flattened array:", flattened)
print("The original does not change because flatten() returns a copy.")

print("\nj. Shape:", matrix.shape)
print("Dimensions (ndim):", matrix.ndim)
print("Size:", matrix.size)
print("Data type (dtype):", matrix.dtype)
```

## Q4. Cognitive-Assistive System and Ordinary Least Squares


```python
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
], dtype=float)

y = np.array([40, 65, 30, 85], dtype=float)

print("X:\n", X)
print("\ny:", y)

print("\na. Shape of X:", X.shape)
print("Dimensions of X:", X.ndim)

XT = X.T
print("\nb. X.T:\n", XT)
print("\nTranspose changes rows into columns and columns into rows.")

XTX = XT @ X
print("\nc. X.T @ X:\n", XTX)

det = np.linalg.det(XTX)
print("\nd. Determinant of X.T @ X:", det)

if not np.isclose(det, 0):
    inverse_XTX = np.linalg.inv(XTX)
    print("Inverse of X.T @ X:\n", inverse_XTX)
else:
    print("X.T @ X is singular, so its regular inverse cannot be calculated.")


if not np.isclose(det, 0):
    beta = np.linalg.inv(XTX) @ XT @ y
else:
    beta = np.linalg.pinv(XTX) @ XT @ y

print("\ne. OLS coefficients (beta):", beta)

print("\nf. Each coefficient shows how the predicted assistance score changes")
print("when the corresponding feature changes by 1 unit, while the other")
print("features are kept fixed.")
print("beta[0] -> Sleep hours coefficient")
print("beta[1] -> Activity level coefficient")
print("beta[2] -> Stress level coefficient")


new_user = np.array([5, 40, 7], dtype=float)
predicted_score = new_user @ beta

print("\ng. New user:", new_user)
print("Predicted assistance score:", predicted_score)
```
