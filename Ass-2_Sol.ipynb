```
L = [10, 0, 20, 40, 10, 70, 0, 0, 0, 80]
scores = tuple(L[:8])

print("Tuple:", scores)
```

```
print("Highest:", max(scores))
print("Highest Index:", scores.index(max(scores)))
print("Lowest:", min(scores))
print("Lowest Count:", scores.count(min(scores)))
```

```
rev = list(scores[::-1])
print("Reversed List:", rev)
```

```
x = int(input("Enter score: "))
if x in scores:
    print("Index:", scores.index(x))
else:
    print("Not Present")
```

```
try:
    scores[0] = 100
except TypeError as e:
    print("Error:", e)
```

```
roll = 1024170008
```

```
digits = [int(i) for i in str(roll)]
```

```
A = {i * 7 for i in digits}
B = {i * 9 for i in digits}

print("Set A:", A)
print("Set B:", B)
```

```
print("Union:", A.union(B))
```

```
print("Intersection:", A.intersection(B))
```

```
print("A - B:", A.difference(B))
print("B - A:", B.difference(A))
```

```
print("Symmetric Difference:", A.symmetric_difference(B))
```

```
print("A is subset of B:", A.issubset(B))
print("B is superset of A:", B.issuperset(A))
```

```
x = int(input("Enter value to remove: "))
A.discard(x)
print("Updated Set A:", A)
```

```
my_dict = {
    "name": "Prabhjot Singh",
    "roll_no": "1024170008",
    "branch": "CSE",
    "age": 20,
    "city": "Patiala"
}

print("Original Dictionary:", my_dict)
```

```
my_dict["location"] = my_dict.pop("city")
print("After Renaming:", my_dict)
```

```
my_dict["cgpa"] = 8.5
print("After Adding CGPA:", my_dict)
```

```
my_dict["age"] += 1
print("After Updating Age:", my_dict)
```

```
d1 = my_dict.copy()
print("Removed:", d1.pop("branch"))
print("Using pop():", d1)
```

```
d2 = my_dict.copy()
del d2["branch"]
print("Using del:", d2)
```

```
for key, value in my_dict.items():
    print(key, "->", value)
```

```
if "email" in my_dict:
    print(my_dict["email"])
else:
    print("Email key not found.")

```

```

friend_dict = {
    "name": "Rahul",
    "roll_no": "1024179999",
    "branch": "CSE",
    "age": 21,
    "city": "Delhi"
}

merged = {**my_dict, **friend_dict}
print("Merged Dictionary:", merged)
```
