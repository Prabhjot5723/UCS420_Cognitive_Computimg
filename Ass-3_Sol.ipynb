## Q6. Create the employees dataset

```python
import pandas as pd

data = {
    "Employee_ID": [101, 102, 103, 104, 105],
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Edward"],
    "Department": ["HR", "IT", "IT", "Marketing", "Sales"],
    "Age": [29, 34, 41, 28, 38],
    "Salary": [50000, 70000, 65000, 55000, 60000],
    "Years_of_Experience": [4, 8, 10, 3, 12],
    "Joining_Date": ["2020-03-15", "2017-07-19", "2013-06-01", "2021-02-10", "2010-11-25"],
    "Gender": ["Female", "Male", "Male", "Female", "Male"],
    "Bonus": [5000, 7000, 6000, 4500, 5000],
    "Rating": [4.5, 4.0, 3.8, 4.7, 3.5]
}

df = pd.DataFrame(data)

# Save the sample dataset as employees.csv
df.to_csv("employees.csv", index=False)

display(df)
```

### Q6(e). Calculate the required statistics

```python
print("i. Average salary:", df["Salary"].mean())
print("ii. Total bonus:", df["Bonus"].sum())
print("iii. Youngest employee's age:", df["Age"].min())
print("iv. Highest performance rating:", df["Rating"].max())
```

### Q6(f). Sort the DataFrame by Salary in descending order

```python
df = df.sort_values(by="Salary", ascending=False)
display(df)
```

### Q6(g). Add Performance category

```python
def performance_category(rating):
    if rating >= 4.5:
        return "Excellent"
    elif rating >= 4.0:
        return "Good"
    else:
        return "Average"

df["Performance"] = df["Rating"].apply(performance_category)
display(df[["Name", "Rating", "Performance"]])
```

### Q6(h). Identify missing values

```python
display(df.isnull().sum())
```

### Q6(i). Rename Employee_ID column to ID

```python
df = df.rename(columns={"Employee_ID": "ID"})
display(df)
```

### Q6(j). Find employees based on experience and department

```python
print("i. Employees with more than 5 years of experience:")
display(df[df["Years_of_Experience"] > 5])

print("ii. Employees in the IT department:")
display(df[df["Department"] == "IT"])
```

### Q6(k). Add Tax column (10% of Salary)

```python
df["Tax"] = df["Salary"] * 0.10
display(df[["ID", "Name", "Salary", "Tax"]])
```

### Q6(l). Save the modified DataFrame to a new CSV file

```python
df.to_csv("employees_modified.csv", index=False)
print("Modified DataFrame saved as employees_modified.csv")
```
